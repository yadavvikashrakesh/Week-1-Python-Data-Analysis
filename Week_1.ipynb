```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('D:/gen AI/Deep Learning/code/AI_ML_Internship_Task/Data/customer_churn_15000.csv')
```

```python
numerical=data.select_dtypes(['number']).columns
```

```python
numerical
```

```python
categorical=data.select_dtypes(['object','category']).columns
```

```python
categorical
```

```python
for col in numerical:
    data[col]=data[col].fillna(data[col].median)
```

```python
for col in categorical:
    data[col]=data[col].fillna(data[col].mode()[0])
```

```python
data.describe()
```

```python
plt.bar(data['city'],data['age'])
plt.title(' city age Trend')
plt.xticks(rotation=90)
plt.xlabel('city')
plt.ylabel('age')

plt.grid()
plt.show()
```

```python
plt.scatter(data['state'],data['support_calls'])
plt.xticks(rotation=90)
plt.title('state vs support calls')
plt.xlabel('state')
plt.ylabel('support calls')
```
